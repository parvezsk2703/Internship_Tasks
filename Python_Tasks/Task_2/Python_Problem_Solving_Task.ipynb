##### Observe: Input is a sentence.

```python
# Q15.
num = [1,2,3,4,5]
eod = {}
for n in num:
    if n % 2 == 0:
        eod[n] = "Even"
    else:
        eod[n] = "Odd"
print(eod)
# Additional test case
num = [1,8,7,4,5]
eod = {}
for n in num:
    if n % 2 == 0:
        eod[n] = "Even"
    else:
        eod[n] = "Odd"
print(eod)
```
