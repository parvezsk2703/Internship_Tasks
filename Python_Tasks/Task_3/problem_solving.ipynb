## Q1. Find the Frequency of Each Word in a Sentence

### Pseudocode
1. Take a sentence.
2. Split it into words.
3. Create an empty dictionary.
4. Count each word and store its frequency.
5. Return the dictionary.

### Expected output (given example)
`{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}`

### Additional test case expected output
`{'cat': 3, 'dog': 1}`

```python
def word_freq(sentence):
    freq = {}
    for word in sentence.split():
        freq[word] = freq.get(word, 0) + 1
    return freq

# Given example
print(word_freq("python is easy and python is powerful"))

# Additional test case
print(word_freq("cat dog cat cat"))
```

## Q2. Find the Frequency of Each Character in a Word

### Pseudocode
1. Take a word.
2. Create an empty dictionary.
3. Visit each character and increase its count.
4. Return the dictionary.

### Expected output (given example)
`{'b': 1, 'a': 3, 'n': 2}`

### Additional test case expected output
`{'h': 1, 'e': 1, 'l': 2, 'o': 1}`

```python
def char_freq(word):
    freq = {}
    for char in word:
        freq[char] = freq.get(char, 0) + 1
    return freq

# Given example
print(char_freq("banana"))

# Additional test case
print(char_freq("hello"))
```

## Q3. Find the First Non-Repeating Character in a Word

### Pseudocode
1. Count each character.
2. Visit characters in original order.
3. Return the first character with count 1.
4. If none exists, return None.

### Expected output (given example)
`'w'`

### Additional test case expected output
`None`

```python
def first_non_repeating(word):
    freq = char_freq(word)
    for char in word:
        if freq[char] == 1:
            return char
    return None

# Given example
print(first_non_repeating("swiss"))

# Additional test case
print(first_non_repeating("aabb"))
```

## Q4. Check Whether Two Strings Are Anagrams

### Pseudocode
1. Count the characters in both strings.
2. Compare the two frequency dictionaries.
3. Return True if they match; otherwise return False.

### Expected output (given example)
`True`

### Additional test case expected output
`False`

```python
def are_anagrams(s1, s2):
    return char_freq(s1) == char_freq(s2)

# Given example
print(are_anagrams("listen", "silent"))

# Additional test case
print(are_anagrams("hello", "world"))
```

## Q5. Find the First Repeating Character in a String

### Pseudocode
1. Create an empty set of seen characters.
2. Visit each character from left to right.
3. If it was seen before, return it.
4. Otherwise add it to the set.
5. If none repeats, return None.

### Expected output (given example)
`'r'`

### Additional test case expected output
`None`

```python
def first_repeating(word):
    seen = set()
    for char in word:
        if char in seen:
            return char
        seen.add(char)
    return None

# Given example
print(first_repeating("programming"))

# Additional test case
print(first_repeating("abcdef"))
```

## Q6. Move All Zeros to the End of a List

### Pseudocode
1. Collect all non-zero values in order.
2. Collect all zeros.
3. Join the two lists and return the result.

### Expected output (given example)
`[3, 1, 2, 4, 0, 0, 0]`

### Additional test case expected output
`[1, 2, 0]`

```python
def move_zeros(nums):
    non_zeros = [num for num in nums if num != 0]
    zeros = [num for num in nums if num == 0]
    return non_zeros + zeros

# Given example
print(move_zeros([0, 3, 0, 1, 2, 0, 4]))

# Additional test case
print(move_zeros([1, 0, 2]))
```

## Q7. Find the Second Highest Distinct Number in a List

### Pseudocode
1. Remove duplicate values.
2. Sort the unique values from largest to smallest.
3. Return the second value if it exists; otherwise return None.

### Expected output (given example)
`30`

### Additional test case expected output
`None`

```python
def second_highest(nums):
    unique_nums = sorted(set(nums), reverse=True)
    if len(unique_nums) < 2:
        return None
    return unique_nums[1]

# Given example
print(second_highest([10, 20, 40, 30, 40, 20]))

# Additional test case
print(second_highest([5, 5, 5]))
```

## Q8. Remove Duplicates Without Changing the Original Order

### Pseudocode
1. Create an empty result list and set.
2. Visit each value.
3. Add it to the result only if it has not been seen.
4. Return the result.

### Expected output (given example)
`[4, 2, 5, 7]`

### Additional test case expected output
`[1, 2, 3]`

```python
def remove_duplicates(nums):
    result = []
    seen = set()
    for num in nums:
        if num not in seen:
            result.append(num)
            seen.add(num)
    return result

# Given example
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Additional test case
print(remove_duplicates([1, 1, 2, 3, 2]))
```

## Q9. Find the Missing Number from 1 to N

### Pseudocode
1. Calculate the expected sum from 1 to N.
2. Calculate the sum of the given list.
3. Subtract the list sum from the expected sum.
4. Return the missing number.

### Expected output (given example)
`3`

### Additional test case expected output
`2`

```python
def missing_number(nums, n):
    return n * (n + 1) // 2 - sum(nums)

# Given example
print(missing_number([1, 2, 4, 5, 6], 6))

# Additional test case
print(missing_number([1, 3, 4], 4))
```

## Q10. Find Two Numbers Whose Sum Equals a Target

### Pseudocode
1. Scan the list from left to right.
2. For each number, check earlier numbers for a matching sum.
3. Return the first qualifying pair.
4. If no pair exists, return an empty list.

### Expected output (given example)
`[2, 7]`

### Additional test case expected output
`[2, 4]`

```python
def two_sum_first(nums, target):
    seen = []
    for num in nums:
        need = target - num
        for earlier in seen:
            if earlier == need:
                return [earlier, num]
        seen.append(num)
    return []

# Given example
print(two_sum_first([2, 7, 11, 15], 9))

# Additional test case
print(two_sum_first([3, 2, 4], 6))
```

## Q11. Find the Number Closest to Zero

### Pseudocode
1. Compare numbers by their absolute values.
2. If tied, choose the positive number.
3. Return the selected number.

### Expected output (given example)
`2`

### Additional test case expected output
`-1`

```python
def closest_to_zero(nums):
    return min(nums, key=lambda num: (abs(num), -num))

# Given example
print(closest_to_zero([-8, -3, 2, 5, -2]))

# Additional test case
print(closest_to_zero([-4, -1, 3]))
```

## Q12. Find Common Elements Between Two Lists

### Pseudocode
1. Check which values from the first list exist in the second list.
2. Keep each common value only once.
3. Preserve the order from the first list.
4. Return the result.

### Expected output (given example)
`[3, 4, 5]`

### Additional test case expected output
`[2, 1]`

```python
def common_elements(list1, list2):
    second_set = set(list2)
    result = []
    seen = set()
    for num in list1:
        if num in second_set and num not in seen:
            result.append(num)
            seen.add(num)
    return result

# Given example
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Additional test case
print(common_elements([2, 2, 3, 1], [2, 1]))
```

## Q13. Rotate a List to the Right by K Positions

### Pseudocode
1. If the list is empty, return an empty list.
2. Reduce K using the list length.
3. Move the last K elements to the front.
4. Return the rotated list.

### Expected output (given example)
`[4, 5, 1, 2, 3]`

### Additional test case expected output
`[3, 1, 2]`

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k else nums[:]

# Given example
print(rotate_right([1, 2, 3, 4, 5], 2))

# Additional test case
print(rotate_right([1, 2, 3], 1))
```

## Q14. Find the Maximum Sum of Two Consecutive Elements

### Pseudocode
1. Add each pair of neighboring elements.
2. Compare all pair sums.
3. Return the largest sum.

### Expected output (given example)
`14`

### Additional test case expected output
`12`

```python
def max_consecutive_sum(nums):
    return max(nums[i] + nums[i + 1] for i in range(len(nums) - 1))

# Given example
print(max_consecutive_sum([4, 7, 2, 9, 5]))

# Additional test case
print(max_consecutive_sum([1, 10, 2]))
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

### Pseudocode
1. Check each pair of distinct list positions.
2. Keep pairs whose sum equals the target.
3. Sort the numbers inside each pair.
4. Remove duplicate pairs and sort the result.

### Expected output (given example)
`[[1, 8], [2, 7], [4, 5]]`

### Additional test case expected output
`[[1, 4], [2, 3]]`

```python
def all_target_pairs(nums, target):
    pairs = set()
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.add(tuple(sorted((nums[i], nums[j]))))
    return [list(pair) for pair in sorted(pairs)]

# Given example
print(all_target_pairs([2, 4, 3, 5, 7, 8, 1], 9))

# Additional test case
print(all_target_pairs([1, 2, 3, 4], 5))
```
